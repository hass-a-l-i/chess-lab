```python
def make_move(piece, from_sq, to_sq, board):
    from_mask = 1 << from_sq
    to_mask = 1 << to_sq

    if not (piece & from_mask):
        raise AssertionError(f"There is no white pawn on {from_sq}")
    
    piece &= ~from_mask
    piece |= to_mask
    return piece
```
