```python
t = list(range(0, 64)) 
board = [t[i:i + 8] for i in range(0, 64, 8)] 
board.reverse() 

for row in board:
    print(" ".join(f"{num:2d}" for num in row))
```

```python
def visualise(board, moves):
    from_squares = {move.from_sq for move in moves}
    to_squares = {move.to_sq for move in moves}

    s = ""
    s += "  ---------------------------\n"

    for rank in range(7, -1, -1):
        row = []

        for file in range(8):
            square = rank * 8 + file

            if square in to_squares:
                row.append("[]")  # available destination
            elif square in from_squares:
                row.append("<>")  # selected/movable piece
            else:
                row.append(board.piece_finder(square))

        s += f"{rank + 1} | {' '.join(row)} | \n"

    s += "  ---------------------------\n"
    s += "     a  b  c  d  e  f  g  h\n"

    return s
```

```python
m = MoveGen(tst_board)
t = visualise(tst_board, m.pawn())
print(t)
print(tst_board)
```
