```python
from_square = 12  # e2
to_square = 28    # e4

from_mask = 1 << from_square
to_mask = 1 << to_square

if b.white_pawns & from_mask:
    print("There is a white pawn on e2")
else:
    print("There is no white pawn on e2")

b.white_pawns & from_mask
b.white_pawns &= ~from_mask
b.white_pawns |= to_mask

if b.white_pawns & from_mask:
    print("There is a white pawn on e2")
else:
    print("There is no white pawn on e2")
```

```python
def print_bitboard(bitboard: int) -> None:
    print("    a b c d e f g h")
    print("  +-----------------+")

    for rank in range(7, -1, -1):
        row = []

        for file in range(8):
            square = rank * 8 + file
            mask = 1 << square

            row.append("1" if bitboard & mask else ".")

        print(f"{rank + 1} | {' '.join(row)} | {rank + 1}")

    print("  +-----------------+")
    print("    a b c d e f g h")

print_bitboard(b.occupied)
```
